```python
import pandas as pd
import numpy as np
```

```python
# Task 1

s1_1 = pd.Series([10, 20, 30, 40, 50])

np_array1_2 = np.random.randint(1, 101, size=(3, 4))

df1_3 = pd.DataFrame(
    np_array1_2,
    columns=['A', 'B', 'C', 'D']
)
print("Series:")
print(s1_1)
print("\nNumPy Array:")
print(np_array1_2)
print("\nDataFrame:")
print(df1_3)
```

```python
# Task 2: Synthetic Data Generation

df_sales = pd.DataFrame({
    'Date': pd.date_range(start='2023-01-01',
                          end='2023-04-10',
                          periods=100),
    'Region': np.random.choice(
        ['East', 'West', 'North', 'South'],
        size=100
    ),
    'Product': np.random.choice(
        ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
        size=100
    ),
    'Sales': np.random.randint(100, 1001, size=100),
    'Quantity': np.random.randint(1, 11, size=100)
})

sales_nan_idx = np.random.choice(df_sales.index,
                                 size=5,
                                 replace=False)

df_sales.loc[sales_nan_idx, 'Sales'] = np.nan

qty_nan_idx = np.random.choice(df_sales.index,
                               size=3,
                               replace=False)

df_sales.loc[qty_nan_idx, 'Quantity'] = np.nan

print(df_sales.head())
```

```python
# Task 3

df3_1 = df_sales[['Region', 'Sales']]

df3_2 = df_sales[df_sales['Product'] == 'Laptop']

df3_3 = df_sales[
    (df_sales['Region'] == 'East') &
    (df_sales['Sales'] > 500)
]

monitor_rows = df_sales[df_sales['Product'] == 'Monitor']

first_monitor_index = monitor_rows.index[0]

val3_4 = df_sales.loc[first_monitor_index, 'Sales']

print("\ndf3_1:")
print(df3_1.head())
print("\ndf3_2:")
print(df3_2.head())
print("\ndf3_3:")
print(df3_3.head())
print("\nFirst Monitor Sales value:", val3_4)
```

```python
# Task 4: Data Cleaning

missing_counts4_1 = df_sales.isnull().sum()

df4_2_filled = df_sales.copy()

sales_mean = df4_2_filled['Sales'].mean()

df4_2_filled['Sales'] = (
    df4_2_filled['Sales']
    .fillna(sales_mean)
)
df4_3_dropped = df_sales.dropna()
print("Missing values in each column:")
print(missing_counts4_1)
print("\nDataFrame with Sales NaNs filled:")
print(df4_2_filled.head())
print("\nDataFrame after dropping NaNs:")
print(df4_3_dropped.head())
```

```python
# Task 5: Data Manipulation
df_sales['Total_Revenue'] = (
    df_sales['Sales'] * df_sales['Quantity']
)
df5_2_sorted = df_sales.sort_values(
    by=['Date', 'Sales'],
    ascending=[True, False]
)
df5_3_grouped_region = (
    df_sales
    .groupby('Region')[['Sales', 'Quantity']]
    .sum()
)
df5_4_grouped_product = (
    df_sales
    .groupby('Product')['Sales']
    .mean()
)
print("\nData with Total_Revenue:")
print(df_sales.head())
print("\nSorted DataFrame:")
print(df5_2_sorted.head())
print("\nGrouped by Region:")
print(df5_3_grouped_region)
print("\nAverage Sales by Product:")
print(df5_4_grouped_product)
```
