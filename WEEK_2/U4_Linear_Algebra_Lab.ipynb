```python
import numpy as np

def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])), # pair 1
    (np.array([1, 2, 3]), np.array([3, 2, 1])), # pair 2
    (np.array([2, 0, 0]), np.array([0, 5, 0])), # pair 3
]
for i, (u, v) in enumerate(pairs, start=1):
    l1_norm = np.linalg.norm(u, ord=1)
    l2_norm = np.linalg.norm(u, ord=2)
    cos_sim = cosine_similarity(u, v)
    print(f"Pair {i}:")
    print(f"  L1 Norm of u: {l1_norm}")
    print(f"  L2 Norm of u: {l2_norm}")
    print(f"  Cosine Similarity: {cos_sim}")
```

```python
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])

print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))
print('A symmetric?', np.allclose(A, A.T))
theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
import numpy as np

M = np.array([[4., 2.], [2., 3.]])
P = np.array([[0., -1.], [1., 0.]])
M_inv = np.linalg.inv(M)
M_trace = np.trace(M)
print("Inverse of M:\n", M_inv)
print("Trace of M:", M_trace)
print("-" * 40)
I = np.eye(M.shape[0])
is_identity = np.allclose(M @ M_inv, I)
print("Is M @ inverse(M) equal to the Identity matrix?", is_identity)
print("-" * 40)
is_symmetric = np.allclose(M, M.T)
is_orthogonal = np.allclose(P.T @ P, np.eye(P.shape[0]))

print("Is M symmetric?", is_symmetric)
print("Is P orthogonal?", is_orthogonal)
```

```python
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])
t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])
scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])
fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
tri = np.array([[0, 1, 0.5], [0, 0, 1.0]])
def close_loop(arr):
    return np.hstack([arr, arr[:, [0]]])
S = np.array([[2.0, 0.0], [0.0, 0.5]])
theta = np.radians(45)
c, s = np.cos(theta), np.sin(theta)
R = np.array([[c, -s], [s, c]])
tri_transformed = R @ S @ tri
tri_closed = close_loop(tri)
tri_trans_closed = close_loop(tri_transformed)
plt.figure(figsize=(6, 6))
plt.plot(
    tri_closed[0],
    tri_closed[1],
    "b-o",
    label="Original Triangle",
    linewidth=2,
)
plt.plot(
    tri_trans_closed[0],
    tri_trans_closed[1],
    "r-s",
    label="Transformed Triangle",
    linewidth=2,
)
plt.axhline(0, color="black", linewidth=0.5, linestyle="--")
plt.axvline(0, color="black", linewidth=0.5, linestyle="--")
plt.grid(True, linestyle=":", alpha=0.6)
plt.axis("equal")
plt.legend()
plt.title("Triangle Transformations: Scaling & Rotation")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
A = np.array([[2., 0.],
              [0., 3.]])
vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
import numpy as np
import numpy.linalg as la

C = np.array([[4., 1.], [2., 3.]])
# 1. eigenvalues and eigenvectors
eigenvalues, eigenvectors = la.eig(C)
# 2. print eigenvalues
print("Eigenvalues:", eigenvalues)
# 3. verify C v = lambda v for the first eigenvector
lam = eigenvalues[0]
v = eigenvectors[:, 0]  # First column corresponds to the first eigenvector
is_close = np.allclose(C @ v, lam * v)
print("Verification for first eigenvector:", is_close)
```

```python
A = np.array([[ 2.,  1., -1.],
              [-3., -1.,  2.],
              [-2.,  1.,  2.]])
b = np.array([8., -11., -3.])
x = la.solve(A, b)
print('Solution x:', x)
print('Rank of A :', la.matrix_rank(A))
print('Full rank -> a unique solution exists')
```

```python
import numpy as np
A2 = np.array([[3., 2.], [1., 4.]])
b2 = np.array([7., 9.])
cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])
x = np.linalg.solve(A2, b2)
print("x:", x)
rank = np.linalg.matrix_rank(A2)
print("Rank of A2:", rank)
def cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
sim_cat_dog = cosine_similarity(cat, dog)
sim_cat_car = cosine_similarity(cat, car)
print(f"Cosine similarity (cat vs dog): {sim_cat_dog:.4f}")
print(f"Cosine similarity (cat vs car): {sim_cat_car:.4f}")
if sim_cat_dog > sim_cat_car:
    print("cat vs dog is more similar.")
else:
    print("cat vs car is more similar.")
```
